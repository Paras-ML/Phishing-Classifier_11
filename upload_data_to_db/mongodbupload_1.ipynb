```python
import pandas as pd
def upload_files_to_mongodb(mongo_client_con_string, database_name, datasets_dir_name):
    # connect to MongoDB Atlas
    client = MongoClient(mongo_client_con_string)
    db = client[database_name]

    for file in os.listdir(datasets_dir_name):
        if file.endswith('.csv'):
            file_name = file.split('.')[0]   # collection name = filename without extension
            collection = db[file_name]

            # read csv
            file_path = os.path.join(datasets_dir_name, file)
            print(f"Reading: {file_path}")
            df = pd.read_csv(file_path)

            # insert records
            data = df.to_dict("records")
            if data:   # insert only if not empty
                collection.insert_many(data)
                print(f"✅ {file_name} uploaded to MongoDB ({len(data)} records)")
            else:
                print(f"⚠️ {file_name} is empty, skipped")

    print("🎉 All CSV files uploaded successfully!")
```
